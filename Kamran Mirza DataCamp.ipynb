```python
# Netflix Movie Analysis (1990s)
# Analyze Netflix movies from the 1990s using Python and pandas.

# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

## Load Dataset
netflix_df = pd.read_csv("netflix_data.csv")

## Most Frequent Movie Duration
nf_movies_1990s = netflix_df[(netflix_df['release_year'] >= 1990) & (netflix_df['release_year'] <= 1999)]
duration = int(nf_movies_1990s['duration'].mode()[0])
print(f'The most frequent movie duration in the 1990s is: {duration} minutes')

## Short Action Movies (<90 min)
short_movie_count = netflix_df[
    (netflix_df['type'] == 'Movie') &
    (netflix_df['genre'] == 'Action') &
    (netflix_df['release_year'] >= 1990) &
    (netflix_df['release_year'] <= 1999) &
    (netflix_df['duration'] < 90)
]
short_movie_count = len(short_movie_count)
print(f'There were {short_movie_count} short action movies released in the 1990s')
```
